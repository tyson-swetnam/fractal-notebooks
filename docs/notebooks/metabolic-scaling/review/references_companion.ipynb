```python
rows = [
  {'cite':'Rubner 1883', 'doi':''},
  {'cite':'Kleiber 1932', 'doi':''},
  {'cite':'Hemmingsen 1960', 'doi':'XX'},
  {'cite':'Bradbury & Reichelt 1983', 'doi':'10.1016/0304-3800(94)00177-4'},
  {'cite':'West et al. 1999', 'doi':'10.1126/science.276.5309.122'},
  {'cite':'West 1999', 'doi':'10.1126/science.284.5420.1677'},
  {'cite':'Brown et al. 2000', 'doi':'10.1073/pnas.97.11.6242'},
  {'cite':'Loehle & Li 1996', 'doi':'10.1016/0304-3800(94)00177-4'},
  {'cite':'West et al. 2010', 'doi':'10.1038/s41586-019-0976-6'},
  {'cite':'Banavar, Maritan & Rinaldo 1999', 'doi':''},
  {'cite':'Calder 1996', 'doi':''},
]
audited = u.audit_citations(rows)
for r in audited:
    tag = ('  FLAG: ' + '; '.join(r['flags'])) if r['flags'] else '  ok'
    print(f"{r['cite']:<34} {r['doi'] or '(none)':<40}{tag}")
```
